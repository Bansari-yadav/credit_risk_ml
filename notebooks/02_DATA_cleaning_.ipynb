```python
import pandas as pd
import numpy as np
```

```python
# Load raw dataset
file_path = r"D:\Work - 2026\credit-risk-ml\data\raw\default of credit card clients.xls"
df = pd.read_excel(file_path, header=1)

print("Shape:", df.shape)
df.head()
```

```python
# Check column names
print(df.columns.tolist())
```

```python
# Rename target column
df = df.rename(columns={"default payment next month": "default_payment_next_month"})

print(df.columns.tolist())
```

```python
# Basic info
df.info()
```

```python
# Missing values
print(df.isnull().sum())
```

```python
# Duplicate rows
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Drop ID column
if "ID" in df.columns:
    df = df.drop(columns=["ID"])

print("Shape after dropping ID:", df.shape)
df.head()
```

```python
# Check category values
print("SEX unique values:", sorted(df["SEX"].unique()))
print("EDUCATION unique values:", sorted(df["EDUCATION"].unique()))
print("MARRIAGE unique values:", sorted(df["MARRIAGE"].unique()))
```

```python
# Clean EDUCATION and MARRIAGE
df["EDUCATION"] = df["EDUCATION"].replace([0, 5, 6], 4)
df["MARRIAGE"] = df["MARRIAGE"].replace(0, 3)

print("Cleaned EDUCATION values:", sorted(df["EDUCATION"].unique()))
print("Cleaned MARRIAGE values:", sorted(df["MARRIAGE"].unique()))
```

```python
# Summary statistics
df.describe()
```

```python
# Target distribution
print(df["default_payment_next_month"].value_counts())
print(df["default_payment_next_month"].value_counts(normalize=True))
```

```python
# Final check
print("Final shape:", df.shape)
df.head()
```

```python
# Save cleaned dataset
output_path = r"D:\Work - 2026\credit-risk-ml\data\processed\credit_default_clean.csv"
df.to_csv(output_path, index=False)

print(f"Cleaned dataset saved at: {output_path}")
```
